# Whisper-large-v3-turbo — Fastest open ASR

Uses `faster-whisper` (CTranslate2 backend). 4-8x faster than `openai-whisper` reference impl.

Model: `openai/whisper-large-v3-turbo` (HuggingFace). Auto-downloads first run.

```python
from faster_whisper import WhisperModel
import time

model = WhisperModel('large-v3-turbo', device='cuda', compute_type='float16')
print('Model loaded.')
```

```python
# Drop an audio file into /workspace/data/ then run:
AUDIO = '/workspace/data/sample.wav'

t0 = time.time()
segments, info = model.transcribe(AUDIO, beam_size=5, vad_filter=True)
segments = list(segments)
elapsed = time.time() - t0

print(f'Language: {info.language} ({info.language_probability:.2f})')
print(f'Duration: {info.duration:.1f}s · processed in {elapsed:.1f}s · {info.duration/elapsed:.1f}x realtime')
print('---')
for s in segments[:10]:
    print(f'[{s.start:6.2f} → {s.end:6.2f}] {s.text}')
```
