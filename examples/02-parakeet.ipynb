# NVIDIA Parakeet — Top-leaderboard ASR

`nvidia/parakeet-tdt-1.1b-v2` — Top Open-ASR-Leaderboard. Native NeMo. ~7x realtime on A100.

Reference: https://huggingface.co/nvidia/parakeet-tdt-1.1b-v2

```python
import nemo.collections.asr as nemo_asr
import time

asr_model = nemo_asr.models.ASRModel.from_pretrained(model_name='nvidia/parakeet-tdt-1.1b-v2')
asr_model = asr_model.train(False).cuda()
print('Parakeet loaded.')
```

```python
AUDIO = '/workspace/data/sample.wav'
t0 = time.time()
results = asr_model.transcribe([AUDIO])
print(f'Processed in {time.time()-t0:.2f}s')
print(results[0])
```
