# NVIDIA Canary — Multilingual ASR + Translation

`nvidia/canary-1b` — 4 languages (en/de/fr/es). ASR + translation in one model.

Reference: https://huggingface.co/nvidia/canary-1b

```python
from nemo.collections.asr.models import EncDecMultiTaskModel
canary = EncDecMultiTaskModel.from_pretrained('nvidia/canary-1b')
decode_cfg = canary.cfg.decoding
decode_cfg.beam.beam_size = 1
canary.change_decoding_strategy(decode_cfg)
canary = canary.train(False).cuda()
print('Canary loaded — supports en/de/fr/es ASR + translation.')
```

```python
asr = canary.transcribe(['/workspace/data/sample.wav'], task='asr', source_lang='en', target_lang='en', pnc='yes')
print('ASR:', asr[0])
translated = canary.transcribe(['/workspace/data/sample.wav'], task='s2t_translation', source_lang='en', target_lang='de', pnc='yes')
print('Translated (de):', translated[0])
```
